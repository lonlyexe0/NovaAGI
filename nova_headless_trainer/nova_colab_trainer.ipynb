## 2. Nova Paketini Yükleme ve Çıkarma
Bilgisayarınızdaki `nova_colab_package.zip` dosyasını Colab'a yükleyin. Aşağıdaki hücreyi çalıştırarak dosya seçebilirsiniz (veya sol paneldeki 📁 simgesine sürükleyip bırakabilirsiniz):

```python
import os, zipfile, glob
from google.colab import files

zip_bulundu = [f for f in os.listdir('.') if f.endswith('.zip')]
if not zip_bulundu:
    print("Lütfen 'nova_colab_light.zip' dosyasını seçin:")
    uploaded = files.upload()
    zip_bulundu = list(uploaded.keys())

hedef_zip = zip_bulundu[0] if zip_bulundu else 'nova_colab_light.zip'
if os.path.exists(hedef_zip):
    with zipfile.ZipFile(hedef_zip, 'r') as z:
        z.extractall('.')
    print(f"🎉 {hedef_zip} başarıyla açıldı ve hazır!")
    !ls -la
```
